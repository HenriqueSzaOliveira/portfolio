# Lendo JSON com PySpark

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

df = spark.read.json(
    '/Workspace/Users/henrique.sza.oliveira@gmail.com/GitHub/portfolio/Databricks/files/livros.json'
)
```

```python
df.select(
    *(
        [c for c in df.columns if c != "endereco"] +
        [
            "endereco.estado",
            "endereco.cidade",
            "endereco.bairro",
            "endereco.rua",
            "endereco.numero",
            "endereco.cep"
        ]
    )
).display()
```

# Lendo CSV com SQL

```python
%sql
SELECT * FROM csv.`/Workspace/Users/henrique.sza.oliveira@gmail.com/GitHub/portfolio/Databricks/files/enderecos.csv`

-- Note que o header nao foi aceito e as colunas sao todas strings, vamos corrijir isso com o read_files
```

```python
%sql
SELECT * FROM
    read_files(
        '/Workspace/Users/henrique.sza.oliveira@gmail.com/GitHub/portfolio/Databricks/files/enderecos.csv',
        format => "csv",
        header => true,
        inferSchema => true,
        sep => ","
    )

-- também podemos informar um schema
```

```python
%sql
SELECT * FROM
    read_files(
        '/Workspace/Users/henrique.sza.oliveira@gmail.com/GitHub/portfolio/Databricks/files/enderecos.csv',
        format => "csv",
        header => true,
        schema => ('id int, rua string, numero int, bairro string, cidade string, estado string, cep string'),
        sep => ","
    )
```

# Agora vamos ver CSV no PySpark

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

df = spark.read.csv(
    '/Workspace/Users/henrique.sza.oliveira@gmail.com/GitHub/portfolio/Databricks/files/enderecos.csv',
    header=True,
    sep=',',
    #inferSchema=True,
    schema=('id int, rua string, numero int, bairro string, cidade string, estado string, cep string'),
).display()
```
